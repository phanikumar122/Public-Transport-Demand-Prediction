# 01 — Data Inspection
Load and inspect all three raw datasets.

```python
import sys, zipfile, pathlib
sys.path.insert(0, str(pathlib.Path('.').resolve().parent))
import pandas as pd

ZIP_ROOT = str(pathlib.Path('..').resolve().parent)

with zipfile.ZipFile(f'{ZIP_ROOT}/apsrtc.zip') as z:
    with z.open('APSRTC_Transport_Data.csv') as f:
        apsrtc = pd.read_csv(f)

print(f'APSRTC Shape: {apsrtc.shape}')
apsrtc.head()
```

```python
print('Dtypes:\n', apsrtc.dtypes)
print('\nMissing values:\n', apsrtc.isnull().sum())
```

```python
apsrtc.describe()
```

```python
with zipfile.ZipFile(f'{ZIP_ROOT}/domestic flights.zip') as z:
    with z.open('flights.csv') as f:
        flights = pd.read_csv(f)
print(f'Flights Shape: {flights.shape}')
flights.head()
```

```python
with zipfile.ZipFile(f'{ZIP_ROOT}/irctc.zip') as z:
    with z.open('IRCTC_cleaned.csv') as f:
        railways = pd.read_csv(f)
print(f'Railways Shape: {railways.shape}')
railways.head()
```

## Data Dictionary

### APSRTC (ML TARGET = `passengers`)

| Column | Type | Description |
|--------|------|-------------|
| route | str | Origin–Destination pair |
| bus_type | str | Service class (Volvo AC, Sleeper…) |
| depot | str | Operating depot |
| date | date | Trip date |
| capacity | int | Seat capacity |
| **passengers** | **int** | **TARGET — passengers carried** |
| occupancy_rate | float | % seats filled (POST-TRIP) |
| distance_km | float | Route distance |
| fare_per_passenger | float | Ticket fare |
| revenue | float | Trip revenue (POST-TRIP) |
