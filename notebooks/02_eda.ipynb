# 02 — Exploratory Data Analysis (EDA)
Analyse APSRTC patterns, distributions, correlations.

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path('..').resolve()))
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline

df = pd.read_csv('../data/processed/apsrtc_features.csv')
df['date'] = pd.to_datetime(df['date'], errors='coerce')
print(f'Shape: {df.shape}')
df.head()
```

```python
# Demand distribution
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
axes[0].hist(df['passengers'], bins=30, color='steelblue', edgecolor='white')
axes[0].set_title('Demand Distribution')
axes[0].set_xlabel('Passengers per Trip')
axes[1].boxplot(df['passengers'], vert=False)
axes[1].set_title('Box Plot')
plt.tight_layout()
plt.savefig('../reports/figures/demand_distribution.png', dpi=150)
plt.show()
```

```python
# Average demand by route
route_demand = df.groupby('route')['passengers'].mean().sort_values(ascending=False)
plt.figure(figsize=(14, 6))
route_demand.plot(kind='bar', color='steelblue', alpha=0.85)
plt.title('Average Passengers by Route', fontsize=14, fontweight='bold')
plt.xticks(rotation=45, ha='right')
plt.ylabel('Average Passengers')
plt.tight_layout()
plt.savefig('../reports/figures/demand_by_route.png', dpi=150)
plt.show()
```

```python
# Demand by day of week
dow_demand = df.groupby('dow')['passengers'].mean()
plt.figure(figsize=(10, 5))
plt.bar(range(7), dow_demand.values, color='coral')
plt.xticks(range(7), ['Mon','Tue','Wed','Thu','Fri','Sat','Sun'])
plt.title('Average Demand by Day of Week')
plt.ylabel('Avg Passengers')
plt.tight_layout()
plt.savefig('../reports/figures/demand_by_dow.png', dpi=150)
plt.show()
```

```python
# Monthly trend
monthly = df.resample('ME', on='date')['passengers'].mean()
plt.figure(figsize=(14, 5))
plt.plot(monthly.index, monthly.values, 'bo-', linewidth=2)
plt.fill_between(monthly.index, monthly.values, alpha=0.3)
plt.title('Monthly Average Demand Trend')
plt.tight_layout()
plt.savefig('../reports/figures/demand_trend.png', dpi=150)
plt.show()
```

```python
# Correlation heatmap
nums = ['passengers','capacity','distance_km','fare_per_passenger','lag_1','lag_7','rolling_mean_7']
corr = df[nums].corr()
plt.figure(figsize=(10, 8))
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm', center=0, square=True)
plt.title('Correlation Heatmap')
plt.tight_layout()
plt.savefig('../reports/figures/correlation_heatmap.png', dpi=150)
plt.show()
```

```python
# Weekend vs Weekday
wkd = df.groupby('is_weekend')['passengers'].mean()
plt.figure(figsize=(6, 5))
plt.bar(['Weekday','Weekend'], wkd.values, color=['steelblue','coral'])
plt.title('Average Demand: Weekday vs Weekend')
plt.ylabel('Average Passengers')
for i, v in enumerate(wkd.values):
    plt.text(i, v + 0.3, f'{v:.1f}', ha='center', fontweight='bold')
plt.tight_layout()
plt.savefig('../reports/figures/weekday_vs_weekend.png', dpi=150)
plt.show()
```
